```python
texts = [
    'I love this movie',
    'This film is amazing',
    'I enjoyed the movie',
    'I hate this movie',
    'This film was terrible',
    'I disliked the movie'
]
labels = [1,1,1,0,0,0]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(num_words=1000, oov_token='<OOV>')
tokenizer.fit_on_texts(texts)
```

```python
word_index = tokenizer.word_index
print(word_index)
```

```python
sequences = tokenizer.texts_to_sequences(texts)
sequences
```

```python
#padding (fixed input size)
from tensorflow.keras.preprocessing.sequence import pad_sequences
max_length = 6
X = pad_sequences(sequences, maxlen=max_length, padding='post')
y = labels
import numpy as np
X = np.array(X)
y = np.array(y)
```

```python
from tensorflow.keras.models import Sequential
from keras.layers import Embedding, Flatten, Dense
model = Sequential([
    Embedding(input_dim=1000, output_dim=16, input_length=max_length),
    Flatten(),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
```

```python
history = model.fit(X,y, epochs=30, verbose=2)
```

```python
#Predict on new text

new_reviews = ['I love this film', 'This movie was horrible']

new_sequences = tokenizer.texts_to_sequences(new_reviews)
new_padded = pad_sequences(new_sequences, maxlen=max_length, padding='post')
predictions = model.predict(new_padded)
print(predictions)
```

```python
#Save model weights
model.save_weights('sentiment.weights.h5')
print('Done')
```

```python
#Reusing the saved weights
#Make sure you build the same model architecture

new_model = Sequential([
    Embedding(input_dim=1000, output_dim=16, input_length=max_length),
    Flatten(),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

new_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

new_model.build(input_shape=(None, max_length))
```

```python
new_model.load_weights('sentiment.weights.h5')
```

```python
new_predictions = new_model.predict(new_padded)
print(new_predictions)
```

```python
new_model.save('sentiment_model.keras')
```
